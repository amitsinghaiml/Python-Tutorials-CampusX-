## List Comprehension

List Comprehension provides a concise way of creating lists.


newlist = [expression for item in iterable if condition == True]


Advantages of List Comprehension
- More time-efficient and space-efficient than loops.
- Require fewer lines of code.
- Transforms iterative statement into a formula.

```python
# add 1 to 10 number to a list 

L = []
for i in range(1,11):
    L.append(i)
    print(L)                   # print is inside of loop 
    
```

```python
L = []
for i in range(1,11):
    L.append(i)
print(L)                         # print is outside of loop 
```

```python
L = []
for i in range(1,11):
    L.append(i)
print(L)
```

```python
L = [i for i in range(1,11)]
print(L)

p = [i*3 for i in range(1,11)]
print(p)
```

```python
# scalar mulplication on a vector

v = [2,3,4]
s = -3
# [-6,-9,-12] output should be

[s*i for i in v]
```

```python
# add squares 

L = [1,2,3,4,5]

[i*i for i in L]
[i**2 for i in L]
```

```python
# print all the numbers by 5 in the range of 1 to 50

[i for i in range(1,51) if i%5 == 0]
```

```python
# find all the languages starts whith p 

languages = ['java','php','python','c','javascript']

[languages for languages in languages if languages.startswith('p')]
```

```python
# nested if with list comprehension

a = ['apple','banana','orange','watermelon','kiwi','grapes']

b = ['apple','kiwi','banana']

# add new list from my_fruit and items if the fruit exist in basket and also starts with 'a'

[c for c in b if c in a if c.startswith('a')]
```

```python
# print 3*3 matrix using list comprehension --> nested list comprehension

[[i*j for i in range(1,4)] for j in range(1,4)]
```

```python
# cartesian product ---> List comprehension on 2 list together
L1 = [1,2,3,4]
L2 = [5,6,7,8]

[i*j for i in L1 for j in L2]
```

Zip


The zip() function returns a zip object, which is an iterator of tuples where the first item in each passed iterator is paired together, and then the second item in each passed iterator are paired together.

If the passed iterators have different lengths, the iterator with the least items decides the length of the new iterator.

```python
# write a program to add items of 2 lists indexwise

L1 = [1,2,3,4]
L2 = [-1,-2,-3,-4]

print(list(zip(L1,L2)))

[i+j for i,j in zip(L1,L2)]
```

```python
a = [1,2,3]
b = a

print(a)
print(b)

a.append(4)      # changes in a leads to changes in b 
print(a)
print(b)

# lists are mutable
```

```python
a = [1,2,3]
b = a.copy()  # the copy function does not make the b list to depend on a 

print(a)
print(b)

a.append(4)
print(a)
print(b)
```
