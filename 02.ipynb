```python
# Arthimetic Operators 
print(5+6)

print(5-6)

print(5*6)

print(5/2)

print(345//10) # int division  ((((Divide the numbers and keep only the whole-number part (discard the decimal part).))))

print(5%2) # remainder (MODULUS)

print(5**2) # power

print(345//100)
```

```python
# Relational Operatos

print(4>5)

print(4<5)
  
print(4>=4)
  
print(4==4)
 
print(4!=4)   # '!' means "Not equal to"













```

```python
# Logical operator
print(1 and 0)

print(1 or 0)

print(not 0)
```

```python
# Bitwise operator  (((it depends on bits 0 and 1 )))

# bitwise and            (((it compare the bits value of the individual numbers)))
print(2 & 3)

# bitwise or
print(2|3)

# bitwise xor
print(2 ^ 3)    #(((Rule: Output is 1 when the inputs are different)))

print(~3)       # ((( It flips every bit in the binary representation of a number.)))

print( 4 >> 2)  #(((It shifts the binary bits to the right by the specified number of positions.)))

print(5 >> 2)

 
#((00000100 >> 2------>>>>00000001))
```

```python
# Assignment operator 

# =

# a=2 

a=2

# a =a+2
a %=2

print(a)
```

```python
# Membership operator 
#A membership operator is used to check whether a value exists in a sequence such as a list, string, tuple, set, or dictionary.

# in /not in

print('D' in 'Delhi')

print('0' in "1,2,3,4,5")
```

```python
# guessing game
# genrate a random integer betweem 1 to 100

import random
jackpot=random.randint(1,100)
guess=int(input('guess number'))
counter = 1
while guess != jackpot:
    if guess < jackpot:
        print('guess higher')
    else:
        print('guess lower')
    guess = int(input('again guess'))
    counter += 1
else:
    print('correct guess','attempts',counter)
    

```
