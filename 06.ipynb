*args and **kwargs

*args and **kwargs are special python keyword that are used to pass the variable length of argument to a fuction 

```python
# *args 
#allows us to pass a variable number of (((non keyword argument))) to a fuction 

def multiply(a,b,c):
    return a*b*c
    
```

```python
multiply(2,3,4)
```

```python
def multiplyication(*args):     # the astric is used when multiple number of input are going to be introduced  
    product =  1                # instead of args we can put anything like yupp etcc..
    
    for i in args:
        product = product * i
    print(args)        
    return product
```

```python
multiplyication(3,4,5)
```

```python
multiplyication(1,2,3,4,5,6,7,8,9)
```

```python
# kwargs kw--->keyword
# **kwargs  ** ----> key-value pairs ((dictionary))
# **kwargs allows us to pass any number of keyword arguments.
# Keyword arguments mean that they contain a key-value pair, like a Python dictionary.

def display(**kwargs):
    for (key,value) in kwargs.items():
        
        print(key,'-->',value)
    
    
```

```python
display(india='delhi',srilanka='colombo',nepal='katmandu')
```

```python
def g(y):
    print(x)
    print(x+1)      # LOCAL VARIABLE ---> variable inside of the fuction
             
                    # GLOBAL VARIABLE ---> variable inside of the program 
x = 5
g(x)
print(x)            # from inside the function(local variable ) we can use Global variable but from the program we cannot use local variabe
```

```python
def f(y):
    x = 1      # this x is local varible
    x += 1                              
    print(x)                               # no relation between these two varibles even though it have same name eg here  (x)
x=5            # this x is global variable
f(x)
print(x)
```

```python
def h(y):
    x += 1       # we dont have value of x here so we took the value of x from global variable butttttt we cannot make any changes in that variable the code will break   
x=5
h(x)
print(x)
```

Higher order function 

```python
# example

def square(x):
    return x**2 

def transform(f,L):  # this is a higher order function  bcoz  its input is a FUNCTION

    output = []
    for i in L:
        output.append(f(i))
        
    print(output)

L= [1,2,3,4,5]

transform(square,L)

```

```python
# if we have to find cubes we have to make another function 

def square(x):
    return x**2 

def cube(x):
    return x**3

def transform(f,L):  # this is a higher order function  bcoz  its input is a FUNCTION

    output = []
    for i in L:
        output.append(f(i))
        
    print(output)

L= [1,2,3,4,5]

transform(cube,L)
```

we can do the cube with the help ok lambda it will be more easier

```python
def square(x):
    return x**2 

def transform(f,L):  # this is a higher order function  bcoz  its input is a FUNCTION

    output = []
    for i in L:
        output.append(f(i))
        
    print(output)

L= [1,2,3,4,5]

transform(lambda x:x**3,L)
```
