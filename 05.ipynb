Differnce between LIST and TUPLE
1. syntax []vs()

2. mutability

3. speed tuple are faster than list bcoz tuple are immutable generally immutable are faster in compared to mutable 

4. memory tuple take less memory than list

5. built in functionality list give more function than tuple

6. error prone (list)

7. usability

```python
import time

L = list(range(1000))
T = tuple(range(1000))

start = time.time()
for i in L:
    i
print('list time',time.time()-start)

start = time.time()
for i in T:
    i
print('tuple time',time.time()-start)
```

```python
import sys

L = list(range(1000))
T = tuple(range(1000))

print('List size',sys.getsizeof(L))
print('Tuple size',sys.getsizeof(T))
```

Sets
A set is an unordered collection of items. Every set element is unique (no duplicates) and must be immutable (cannot be changed).

However, a set itself is mutable. We can add or remove items from it.

Sets can also be used to perform mathematical set operations like union, intersection, symmetric difference, etc.

Characterstics:

1.Unordered

2.Mutable

3.No Duplicates

4.Can't contain mutable data types-----> sets inside sets is not possible 

```python
# empty stes 
a = {}  #when we make an empty set like this it becomes dictionaries 
print(a)
print(type(a))

s = set()
print(s)
print(type(s))
```

```python
#1D and 2D
s1 = {1,2,3,4}
print(s1)

s2 = {1,2,3,{4,5}}
print(s2)           # shows error 

#    A set itself is mutable, but its elements must be hashable.
  #   That means the elements you put inside a set must generally be immutable.   # we can put tuple/string/int
```

```python
# homo and hetro
s3 = {1,'hello',4.5,True}  # here True---->1 same and same item cannot be repeated inside of a set 
print(s3)

s4 = {1,4.5,'hello',(1,2,3)}   # sets do not give importance to order we cannot change it 
print(s4)
```

```python
# using type conversion

s5 = set([1,2,3])
print(s5)
```

```python
# duplicates not allowed 
s6 = {1,2,3,4,55,6,6,4}
print(s6)
```

```python
# set can't have mutable item 
s7 = {1,2,{3,4}}
print(s7)     # set contain immutabole data type
```

```python
s1 = {1,2,3}
s2 = {3,2,1}
print(s1 == s2)
```

Set function 

```python
# len/sum /min/max/sorted
s = {3,1,2,4,5,7}
len(s)
sum(s)
min(s)
max(s)
sorted(s,reverse = True)
```

```python
# union /update
s1 = {1,2,3,4,5}
s2 = {4,5,6,7,8,9}

# s1|s2 
s1.union(s2)   # it will combine the input of s1 and s2 

s1.update(s2)  # update will add the input of s2 in s1 

print(s1)
print(s2)

```

```python
# interaction / intersection_update
s1 = {1,2,3,4,5}
s2 = {4,5,6,7,8,9}

s1.intersection_update(s2)
print(s1)
print(s2)
```

```python
# differnce/ difference_update
s1 = {1,2,3,4,5}
s2 = {4,5,6,7,8,9}

s1.difference_update(s2)

print(s1)
print(s2)
```

```python
# symmetric_differnce / symmetric_differnce_udate
s1 = {1,2,3,4,5}
s2 = {4,5,6,7,8,9}

s1.symmetric_difference_update(s2)

print(s1)
print(s2)
```

```python
# isdisjoint

s1 = {1,2,3,4}
s2 = {3,4,5,6}

s1.isdisjoint(s2)     # disjoint sets are those sets which do not have nay item in common 
```

```python
# issubset
s1 = {1,2,3,4,5}
s2 = {3,4,5}

s1.issubset(s2)
```

```python
s2.issubset(s1)
```
