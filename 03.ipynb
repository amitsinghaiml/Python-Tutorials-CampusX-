#Sequence sum

1/1! + 2/2! + 3/3! + ...

```python
# code

n = int(input('enter n'))
result = 0 
fact = 1

for i in range(1,n+1):
    fact = fact * i
    result = result + i/fact
    print(result)
     
```

```python
numbers = [12, 5, 8, 21, 14, 7, 10, 33, 18, 9]

even_count = 0
odd_count = 0

for i in numbers:

    if i % 2 == 0:
        even_count += 1

    else:
        odd_count += 1

print('Total even numbers:', even_count)
print('Total odd numbers:', odd_count)
```

```python
# code 

rows = int(input('no of rows'))

for i in range(1,rows+1):
    for j in range(1,i+1):
        print(j,end = '')    #  (end = '') is used to stop Python from moving to the next line after each print().
    for k in range(i-1,0,-1):
        print(k,end = '' )
        
    print()
```
